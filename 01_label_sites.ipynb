```python
import pandas as pd
import numpy as np
import os

# Project path
PROJECT_DIR = r"C:\GeoSense_Agent"

# Features file
FEATURES_FILE = os.path.join(
    PROJECT_DIR,
    "data",
    "processed",
    "features.csv"
)

print("Looking for:")
print(FEATURES_FILE)

print("\nFile exists:", os.path.exists(FEATURES_FILE))

df = pd.read_csv(FEATURES_FILE)

print("\nDataset shape:", df.shape)
print("\nColumns:")
print(df.columns.tolist())

df.head()
```

```python
df[['dist_road_m', 'dist_hospital_m', 'flood_risk']].describe()
```

```python
print("Number of locations:", len(df))
print("Flood-risk values:")
print(df['flood_risk'].value_counts())
```

```python
# Check the actual Kolkata feature ranges

print("Road distance:")
print(df["dist_road_m"].describe())

print("\nHospital distance:")
print(df["dist_hospital_m"].describe())

print("\nFlood risk:")
print(df["flood_risk"].describe())
```

```python
# Create labels for the Kolkata study area

good_condition = (
    (df["dist_road_m"] <= 100) &
    (df["dist_hospital_m"] <= 1000) &
    (df["flood_risk"] <= 2)
)

bad_condition = (
    (df["dist_road_m"] >= 200) |
    (df["dist_hospital_m"] >= 2000) |
    (df["flood_risk"] >= 8)
)

df["label"] = np.nan

df.loc[good_condition, "label"] = 1
df.loc[bad_condition, "label"] = 0

labelled = df.dropna(subset=["label"]).copy()
labelled["label"] = labelled["label"].astype(int)

print("Total labelled:", len(labelled))
print("\nLabel distribution:")
print(labelled["label"].value_counts())
```

```python
print("Good sites (1):", (labelled["label"] == 1).sum())
print("Bad sites (0):", (labelled["label"] == 0).sum())
```

```python
OUTPUT_FILE = r"C:\GeoSense_Agent\data\processed\labelled_sites.csv"

labelled.to_csv(OUTPUT_FILE, index=False)

print("Saved:", OUTPUT_FILE)
print("Rows:", len(labelled))
```
